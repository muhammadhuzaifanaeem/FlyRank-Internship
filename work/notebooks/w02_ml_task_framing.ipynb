## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

```python
# Each row means one page on a site

import pandas as pd

df = pd.read_csv("https://raw.githubusercontent.com/muhammadhuzaifanaeem/FlyRank-Internship/main/data/raw/content_refresh_anonymized.csv")
df.shape
df[["content_id", "trend_pct", "avg_position", "days_since_last_update", "clicks_last_30d", "clicks_prev_30d"]].head()
```

```python
traffic_decline= (-df['trend_pct'].fillna(0)/100).clip(lower=0, upper=1)
signal_position= df['avg_position'].fillna(df["avg_position"].median()/100).clip(upper=1)
days_update= (df['days_since_last_update'].fillna(0)/365)

df['decay_score']= traffic_decline*0.5 + signal_position*0.3 + days_update*0.2

df[["content_id", "trend_pct", "avg_position", "days_since_last_update", "decay_score"]].sort_values("decay_score", ascending=False).head(10)
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
#A fixed rule (e.g., "refresh anything older than 12 months") ignores interactions between signals, a 2-year-old page with strong current rankings may need no refresh, while a 3-month-old page with a sharp traffic drop needs one urgently. A model can learn which combinations of signals actually predict decay, and weight them based on evidence rather than a guess.
```
